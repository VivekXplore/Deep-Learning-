````python
import torch.nn as nn 
import torch

````

we are making NN that has 5 input feature one bias and output layer 

````python
class Model(nn.Module): # has to be baseclass as nn.module
    def __init__(self,num_feature):
        super().__init__()
        self.linear=nn.Linear(num_feature,1)
        self.sigmoid=nn.Sigmoid()

    def forward(self,features):
    
        z=self.linear(features) # auto calculates z=wx+b we are passing feature to linear layer 
        ypred=self.sigmoid(z)
        return ypred

    def loss(ypred,y):
        criterion=nn.BCELoss()
        return criterion
````

````python
# create dataset 
data=torch.rand(10,5) # data with 10 row and 5 columns.

# define model 
model=Model(data.shape[1]) # provide the no of feature

# fordward pass
model(data)
````

````python
class NeuralNetwork(nn.Module):
    def __init__(self,no_features):
        super().__init__()
        self.network=nn.Sequential(
            nn.Linear(no_features,1), # feature, bias for linear layer 1,
            nn.ReLU(),
            nn.Linear(3,1), # feature,bias for linear layer 2
            nn.Sigmoid()
            )
    def forward(self,feature): # done in traning feature
        output=self.network(feature)
        return output

# insetead of writing like above we will make use of container so we can pass desired number of linear layers too. 
````
