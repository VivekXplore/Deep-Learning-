say we have mathemetical relation y=x2 then we can easily calculate the derivative/slope and also code it 

But say 
y=x2 
z=sin(y) then dervaite is dz/dx=dz/dy x dy/dx ..

- Autograd is Automatic Diffrentiation, core component of pytorch that provide diffrentiation for tensors. 
- Enables gradient calculation , 



```python
# if we have to find derivative of y w.r.t x , y=x**2 

# by finding slope manually we do 

def dy_dx(x):
    return 2 * x  # slope of y= x square is 2x . 
dy_dx(3) # slope at x=3 point 
```

```python
''' now, 
y=x^2 
z=sin(y) 

so here dz/dx= dz/dy * dy/dx cuz z depend on y.

so the slope is = cos(y)*2x 
= cos(x**2)*2x as y=x**2'''

import math
def dz_dx(x):
    return math.cos(x**2) *2*x

dz_dx(4)
```

#### Using pytorch : 

```python
import torch 
# for y=x^2 

# define the tensor first . 

x=torch.tensor(3.0 , requires_grad=True) # we defined scalar . 
# requires_gard=True tells torch that we need gradient for this tensor . 

y= x ** 2 # define the relation 
```

```python
# lets print x and y and see 
print(x)
print(y)
```

x, is 3 as we defined. 

y=9 cuz till now we havent calculated the gradient . so it just did the power operation (fordward pass) so x=3 then x**2 = 9 ..

```python
# so to calc derivativae we do 

y.backward() # calc the gradient
x.grad # to see the grad
```

```python
# now we will use pytorch and do it for: 
'''
y=x^2 
z=sin(y) 

so here dz/dx= dz/dy * dy/dx cuz z depend on y.
''' 

# using torch we can simply 

# define tensor or provide val of x to calc derivative at a point 
x=torch.tensor(4.0, requires_grad=True)
# define relation
y=x **2
z=torch.sin(y)

# calc gradient

z.backward()
x.grad
```

```python
# define input and tensors 

x=torch.tensor(6.7) # input feature
y=torch.tensor(0.0) # acutal 

w=torch.tensor(1.0, requires_grad=True) # weight
b=torch.tensor(0.0, requires_grad=True) # bias

# define relation

z= w*x+b
ypred=torch.sigmoid(z)

def binary_cross_entropy(predection,target):
    return -(target)*torch.log(predection)-(1-target)*torch.log(1-predection)

loss=binary_cross_entropy(ypred,y)
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```

### clearing gradient concept 

There is a problem with backward() is is that we we run backward multiple times the gradient gets accumulated . and we dont want that as due to it we wont be able to find the accurate one .. 


```python
# lets see how it accumulates 
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)  # 4

# same relation , same x value but still it will give 8 instaed of 4 cuz it accumulates 
y = x ** 2

y.backward()
print(x.grad)  # 8
```

```python
# clearnign grad 

x= torch.tensor(2.0, requires_grad=True)
```

```python
y = x ** 2
```

```python
y.backward()
```

```python
x.grad
```

```python
# add clearnign grad code 

x.grad.zero_()

# now if you run from y=x**2 and backward theen it will not accumulate it will clear the gradient 
```
