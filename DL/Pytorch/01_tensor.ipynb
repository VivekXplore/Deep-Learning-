```python
# lets see why we need gpu 

''' 1. What operation will we perform?
Suppose we have two large matrices:
- Matrix A: 10,000 × 10,000
- Matrix B: 10,000 × 10,000
We want to multiply them:

   C=A X B

This is called matrix multiplication.

It involves approximately 1 trillion multiplication-and-addition operations. This is computationally expensive.'''

import torch
import time

# ==========================================
# 1. CHECK AVAILABLE DEVICES
# ==========================================

print("PyTorch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())
print("MPS Available:", torch.backends.mps.is_available())

if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("Apple GPU is available!")
else:
    device = torch.device("cpu")
    print("Apple GPU unavailable. Using CPU.")

# ==========================================
# 2. CREATE LARGE MATRICES
# ==========================================

N = 4096

print("\nMatrix size:", N, "x", N)

A = torch.randn(N, N)
B = torch.randn(N, N)

# ==========================================
# 3. CPU COMPUTATION
# ==========================================

print("\nStarting CPU computation...")

# Warm-up
_ = torch.matmul(A, B)

start = time.perf_counter()

C_cpu = torch.matmul(A, B)

cpu_time = time.perf_counter() - start

print("CPU execution time:", round(cpu_time, 4), "seconds")

# ==========================================
# 4. APPLE GPU COMPUTATION
# ==========================================

if device.type == "mps":

    # Transfer matrices to Apple GPU
    A_gpu = A.to(device)
    B_gpu = B.to(device)

    # Warm-up GPU
    _ = torch.matmul(A_gpu, B_gpu)

    torch.mps.synchronize()

    start = time.perf_counter()

    C_gpu = torch.matmul(A_gpu, B_gpu)

    # Wait for GPU to finish
    torch.mps.synchronize()

    gpu_time = time.perf_counter() - start

    print("Apple GPU execution time:",
          round(gpu_time, 4), "seconds")

    # ======================================
    # 5. COMPARE PERFORMANCE
    # ======================================

    print("\n========== RESULTS ==========")

    print("CPU time:", round(cpu_time, 4), "seconds")
    print("GPU time:", round(gpu_time, 4), "seconds")

    speedup = cpu_time / gpu_time

    print("GPU speedup:", round(speedup, 2), "times")

    # Check whether results are similar
    print(
        "Results match:",
        torch.allclose(C_cpu, C_gpu, rtol=1e-3, atol=1e-3)
    )

else:
    print("\nGPU benchmarking skipped.")
    print("Your Mac does not currently support MPS.")

print("\nBenchmark completed!")
```
